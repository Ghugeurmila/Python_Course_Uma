```python
letter="hey my name is '{}' and I am from {}"
country="India"
name="Uma"
print(letter.format(name, country))
```

```python
letter="hey my name is '{1}' and I am from {0}"
country="India"
name="Uma"
print(letter.format(name, country))
```

```python
print(f"Hey my name is {name} and i am from {country}")
```

```python
txt="i need this {money:.2f} for admission"
print(txt.format(money=23324.5657456756))
```

```python
price=365.234627536
txt=f"i need this {price:.2f} for admission"
print(txt)
```
