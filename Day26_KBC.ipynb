Day27:KBC game

Create a progarm capable of displaying questions to user like KBC ,use list data type to store the question and their answer.Display the final amount the person is taking home after playing the game

```python
kbc=["What is your name",'1-Uma','2-mrunali','3-sharvari','4-anushka',1000]
kbc2=["What is you fav food","1-korean","2-chinese","3-italian",'4-indian',2000]
amt=0
j=0
while(j<1):
    for i in range(5):
      print(kbc[i])
    ans=input("enter your choice ")
    if(ans=='1'):
        amt+=1000
        print("Congratulation😘 you won 1000")
    else:
        print("OOPS☹️you lost")
        print("your total reward is ",amt)
        break
    for i in range(5):
      print(kbc2[i])
    ans=input("Enter your choice ")
    if(ans=='4'):
        amt+=2000
        print("Congratulation😘 you won 2000")
    else:
        print("OOPS☹️you lost")
        print("your total reward is ",amt)
        break
    j+=1
print("Total you won ",amt)
```
