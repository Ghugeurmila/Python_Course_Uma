Methods

```python
from typing import Union
s1={1,2,3,4}
s2={3,4,5,6,8}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1)
print(s2)
s1.update(s2) #update set s1
print(s1)
```

```python
s4=s1.symmetric_difference(s2)
print(s4)
```

```python
s5={1,2,3,4,5}
s6={3,4,5,6,7,8}
s7=s6.symmetric_difference_update(s5)
print(s7)
```

```python
#disjoint = no common element
print(s1.isdisjoint(s4))
```

```python
#superset=universal like set
print(s5.issuperset(s1))
print(s6.issubset(s2))
```

```python
'''remove=remove element but error if that element not in set
  discard->it remove element not give error even if that element is not present
'''
print(s1)
s1.remove(5)
s1.discard(100)
print(s1)
```

```python
#pop->remove random element(random->bacause set is unodered)
print(s1.pop())
```

```python
#del->delete the set
s3={4,2,6,5,2}
s3.clear() #clear all element
print(s3)
del s3 #delete set
print(s3)
```
