```python
a=input("Enter the number")
print(f"Table of {a} is ")

for i in range(11):
  print(f" {a} X {int(a)*i}")

print("some imp line after this")
print("End of program")
```

```python
a=input("Enter the number")
print(f"Table of {a} is ")
try:
  for i in range(11):
    print(f" {a} X {int(a)*i}")
except Exception as r:
  print(r)
print("\n some imp line after this")
print("End of program")
```

```python
try:
  num=int(input("Enter the number"))
  a=[4,5,7,3,1,7]
  print(a[num])
except ValueError:
  print("Value is not interger")
except IndexError:
  print("Index error")
```
