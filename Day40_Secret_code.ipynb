```python
def encrypt(sen):
  output=""
  words=sen.split()
  for i in words:
    if len(i)<4:
      output=output+" "+i[::-1]
    else:
      text=i
      reverser=text[len(text):0:-1]
      output=output+" ert"+text[0]+reverser+"wds"
  print("Message send is :",output)
  #print(output)
  decrypt(output)

def decrypt(sen):
  output=""
  words=sen.split()
  for i in words:
    if len(i)<4:
      output=output+" "+i[::-1]
    else:
      text=i[3:]
      text=text[:-3]
      reversed=text[len(text):0:-1]
      output=output+" "+text[0]+reversed
  print("Message receive as :",output)


sen=input("Enter message to send your friend ")
encrypt(sen)
```
