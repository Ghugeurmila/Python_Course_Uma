```python
# like in java and c++ there is switch case
# there is break use use in compulsory
# here that is not needed
x=int(input("Enter you age: "))
match x:
  case 0:
    print("X is 0")
  case 1:
    print('x is 1')
  #case if if statment
  case _ if x>18 and x<=20:
    print("Ready to drive")
  case _ if x>21 and x<=30:
    print("You are ready to marraige")
  case _ :
    print("You can die")
```
