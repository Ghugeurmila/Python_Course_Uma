Finally Keyword
always execute this code inside finally

```python
try:
  l=[23,54,265,34,64]
  i=input("Enter the index: ")
  print(l[i])
except:
  print("Error is ther")

finally:
  print("I am always execute")
print("I am outside the finally still i am\n always excute")

```

finally is used in function where return is there still some code needed to be run

```python
def func():
  try:
    l=[2,3,5,6,2]
    i=input("enter the index: ")
    print(l[int(i)])
    return 0
  except Exception as e:
    print(e)
    return 1

  finally:
    print("I am always here to run")
    d=input("Enter your name")
    print(f"you {d} are so beautiful")

func()
```
