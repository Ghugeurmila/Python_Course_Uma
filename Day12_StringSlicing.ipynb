String Slicing

```python
name="URMILA KAHDNU GHUGE"
```

```python
print(name[0:6])#last is skiped 6 is skip here
print(name[:6])
print(name[14:])
```

```python
print(len(name))
```

Negative Slicing


```python
print(name[-12:])#last as -1 then -2....
print(name[7:-5])
```
