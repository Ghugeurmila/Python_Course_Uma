For loop wiith else

```python
for i in range(5):
  print(i)
else:
  print("loop end")
```

```python
for i in range(6):
  print(i)
  if(i==4):
    break
else:
  print("only run after loop end \n if break then else dont print")
```

```python
i=0
while(i<6):
  print(i)
  i=i+1
  if(i==4):
    print("Forcefull exit")
    break
else:
  print("print only after loop end")
```
