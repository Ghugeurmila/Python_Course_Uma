Recursion in python

Factorial of n

```python
def fact(n):
  if(n==0 or n==1):
    return 1
  else:
    return n*fact(n-1)

fact(4)
```

Fibonacci series

f(n)=f(n-1)+f(n-2)

```python
def fibo(n):
  if(n==0):
    return 0
  elif(n==1):
    return 1
  else:
    return fibo(n-1)+fibo(n-2)

fibo(6)
```
