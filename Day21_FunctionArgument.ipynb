Function Arguments


```python
def average(a,b):
  print("the avg is ",(a+b)/2)
average(3,2)
```

```python
def average(a=3,b=6):
  print("the avg is ",(a+b)/2)

average()
```

```python
def average(a,b=8):
  print("the avg is ",(a+b)/2)

average(3)
```

```python
def average(a=3,b=1):
  print("the avg is ",(a+b)/2)

average(b=8)
```

```python
def averageNum(*numbers):
  sum=0
  for i in numbers:
    sum+=i
  print("Average is ",sum/len(numbers))
averageNum(2,3,4,2,1)
```
