OS module

```python
import os

```

```python
#create data folder if not exist
if(not os.path.exists("data")):
  os.mkdir("Data")

#create day1 to day100 folder  in data folder
for i in range(0,100):
  os.mkdir(f"Data/Day{i+1}")
```

```python
#rename the folders
for i in range(0,100):
  os.rename(f"Data/Day{i+1}",f"Data/Titorial{i+1}")
```

```python
#print list of folders
folders= os.listdir("Data")
print(folders)

for i in folders:
  print(i)#each folder print one by one above print in mess
```

```python
#print folder and inside fiels too
for i in folders:
  print(folders)
  print(os.listdir(f"Data/{i}"))
```
