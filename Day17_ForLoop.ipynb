For Loop

```python
name="URMILA"
for i in name:
  print(i)
```

```python
#use for for the list
color=['red','yellow','orange','blue']
for i in color:
  print(i)
```

```python
#use for for the list
color=['red','yellow','orange','blue']
for i in color:
  print(i)
  for j in i:
    print(j)
```

```python
for k in range(5):
  print(k)
```
