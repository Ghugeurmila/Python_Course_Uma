Local Vs Global

```python
x=5

def hello():
  x=3
  print(f"value of x is {x}")

print(f"value of x is {x}")
hello()
```

```python
x=10

def he():
  global x  #change the value of global here
  x=3
  y=5
  print(x)
  print(y)

he()
print(x)
```
