break and continue

```python
for i in range(12):
  if(i==10):
    break #skip these entire loop
  print("5x", i+1, "= ",5*(i+1) )
```

```python
for i in range(20):
  if(i%5==0):
    continue
  print("5 X", i,"= ",5*i)
```
