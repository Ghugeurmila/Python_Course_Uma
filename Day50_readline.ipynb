File IO,read,readline

```python
f=open('my_file.txt','r')
while True:
  line=f.readline()
  if not line:
    break
  print(line)
```

```python
f=open('marks.txt','r')
i=0
while True:
  i=i+1
  line= f.readline()
  if not line:
    break
  m1= line.split(",")[0]
  m2=int(line.split(",")[1])#now marks consider here in integer form above in string
  m3=line.split(",")[2]
  print(f"marks of student {i} in Math is {m1}")
  print(f"marks of student {i} in English is {m2}")
  print(f"marks of student {i} in Science is {m3}")

  print(line)
```
